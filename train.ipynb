```python
import network
import emnist_loader
```

```python
# create a network
# OUTPUT 36 Class (0-9 + A-Z)
net = network.Network([784,64,32,36])
```

```python
# load the data
training_data, validation_data, test_data = emnist_loader.load_data_wrapper()
```

```python
# train the network
net.train(training_data, epochs=10, mini_batch_size=10, eta=3.0, test_data=test_data)
```

```python
# Save the trained model
# MODEL EMNIST
net.save("trained_network_emnist.pkl")
```
